```python
total_charges_numeric = pd.to_numeric(
    df['TotalCharges'],
    errors='coerce'
)

invalid_total_charges = df[
    total_charges_numeric.isna()
]

invalid_total_charges[
    ['customerID', 'tenure', 'MonthlyCharges', 'TotalCharges', 'Churn']
]
```

```python
df.loc[
    df['tenure'] == 0,
    ['customerID', 'tenure', 'MonthlyCharges', 'TotalCharges', 'Churn']
]
```

```python
print(
    "Customers with tenure 0 :",
    (df['tenure'] == 0).sum()
)

print(
    "Customers with TotalCharges missing :",
    total_charges_numeric.isna().sum()
)
```

```python
# Convert totalcharge from str to int/float
df['TotalCharges'] = total_charges_numeric

# for customers with tenure 0, no charges have accumulated yet 
df.loc[
    df['tenure'] == 0, 'TotalCharges'
] = 0
```
